**# Task 4: Functions for Reusable Data Tasks**

This notebook demonstrates reusable Python functions for calculations, data transformation, and statistical operations.

**## 1. Creating a Custom Function**

A function is a reusable block of code defined using the `def` keyword.

```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Monika"))
print(greet("Student"))
```

**## 2. Calculating Mean**

This function calculates the average of numerical values using their sum and count.

```python
def calculate_mean(numbers):
    return sum(numbers) / len(numbers) if numbers else 0

marks = [75, 80, 90, 65, 85]
print("Mean:", calculate_mean(marks))
```

**## 3. Finding Maximum and Minimum**

These functions identify the largest and smallest values in a dataset.

```python
def find_maximum(numbers):
    return max(numbers) if numbers else None

def find_minimum(numbers):
    return min(numbers) if numbers else None

values = [12, 45, 7, 89, 23]

print("Maximum:", find_maximum(values))
print("Minimum:", find_minimum(values))
```

**## 4. Calculating Percentage**

This function calculates a student's percentage based on obtained and maximum marks.

```python
def calculate_percentage(obtained, maximum):
    if maximum <= 0:
        return None
    return (obtained / maximum) * 100

percentage = calculate_percentage(425, 500)
print("Percentage:", percentage, "%")
```

**## 5. Data Transformation**

This function transforms a list of numbers by multiplying each value by a given factor.

```python
def transform_data(numbers, factor):
    return [number * factor for number in numbers]

data = [10, 20, 30, 40]
print("Original Data:", data)
print("Transformed Data:", transform_data(data, 2))
```

**## 6. Employee Salary Calculation**

This function calculates an employee's updated salary using a percentage increment.

```python
def updated_salary(salary, increment_percent):
    return salary + (salary * increment_percent / 100)

salary = 25000
print("Updated Salary:", updated_salary(salary, 10))
```

**## 7. Dataset Summary**

This function returns important summary statistics for a numerical dataset.

```python
def summarize_data(numbers):
    if not numbers:
        return {"count": 0, "total": 0, "mean": None}

    return {
        "count": len(numbers),
        "total": sum(numbers),
        "mean": calculate_mean(numbers),
        "maximum": find_maximum(numbers),
        "minimum": find_minimum(numbers)
    }

sales = [1200, 1500, 1800, 1000, 2000]
print(summarize_data(sales))
```
