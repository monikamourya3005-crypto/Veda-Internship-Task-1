```python

# Day 36 - Probability Fundamentals
# Veda Technology Data Science Internship

# Probability measures the likelihood of an event.
# It ranges from 0 (impossible) to 1 (certain).

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Probability Fundamentals")
print("Day 36 - Veda Internship")
```

2 – Basic Probability Calculations

```python

# Probability = Favorable Outcomes / Total Outcomes

# Example 1: Probability of getting Head in a coin toss
favorable_outcomes = 1
total_outcomes = 2

coin_probability = favorable_outcomes / total_outcomes

print("Probability of getting Head:", coin_probability)

# Example 2: Probability of getting an even number on a dice
favorable_outcomes = 3
total_outcomes = 6

even_probability = favorable_outcomes / total_outcomes

print("Probability of getting an even number:", even_probability)

# Example 3: Probability of getting number 6
six_probability = 1 / 6

print("Probability of getting 6:", six_probability)
```

3 – Coin Toss Simulation

```python

# Simulate 1000 coin tosses using NumPy.
# Each toss can result in Head or Tail.

np.random.seed(42)

tosses = np.random.choice(
    ["Head", "Tail"],
    size=1000
)

heads = np.sum(tosses == "Head")
tails = np.sum(tosses == "Tail")

head_probability = heads / len(tosses)
tail_probability = tails / len(tosses)

print("Total Tosses:", len(tosses))
print("Heads:", heads)
print("Tails:", tails)

print("Experimental Probability of Head:", head_probability)
print("Experimental Probability of Tail:", tail_probability)
```

4 – Dice Roll Simulation

```python

# Simulate rolling a six-sided dice 1000 times.
# Dice outcomes range from 1 to 6.

dice_rolls = np.random.randint(1, 7, size=1000)

for number in range(1, 7):
    count = np.sum(dice_rolls == number)
    probability = count / len(dice_rolls)

    print(
        "Number:", number,
        "| Frequency:", count,
        "| Probability:", round(probability, 4)
    )
```

5 – Theoretical vs Experimental Probability

```python

# Theoretical probability of getting Head is 0.5.
# Experimental probability is calculated from actual simulations.

simulation_sizes = [10, 50, 100, 500, 1000, 5000, 10000]

experimental_probabilities = []

for size in simulation_sizes:
    tosses = np.random.choice(
        ["Head", "Tail"],
        size=size
    )

    probability = np.sum(tosses == "Head") / size
    experimental_probabilities.append(probability)

theoretical_probability = 0.5

print("Theoretical Probability:", theoretical_probability)

for size, probability in zip(
    simulation_sizes,
    experimental_probabilities
):
    print(
        "Tosses:", size,
        "| Experimental Probability:", round(probability, 4)
    )
```

6 – Probability Convergence Graph

```python

# As the number of experiments increases,
# experimental probability generally approaches theoretical probability.

plt.figure(figsize=(10, 5))

plt.plot(
    simulation_sizes,
    experimental_probabilities,
    marker='o',
    label='Experimental Probability'
)

plt.axhline(
    y=theoretical_probability,
    linestyle='--',
    label='Theoretical Probability'
)

plt.xlabel("Number of Coin Tosses")
plt.ylabel("Probability")
plt.title("Theoretical vs Experimental Probability")

plt.legend()
plt.grid(True)
plt.show()
```
