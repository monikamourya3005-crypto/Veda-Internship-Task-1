1: Import NumPy and generate random numbers

```python
# Day 13: NumPy Random Sampling
# Import NumPy for random number generation

import numpy as np

# Generate random integers
random_integers = np.random.randint(1, 101, size=10)

# Generate random floating-point numbers
random_floats = np.random.random(5)

print("Random Integers:", random_integers)
print("Random Floating-Point Numbers:", random_floats)
```

2: Random seed and reproducibility

```python
# A fixed seed makes random results reproducible

np.random.seed(42)
sample1 = np.random.randint(1, 101, size=10)

np.random.seed(42)
sample2 = np.random.randint(1, 101, size=10)

print("First Sample:", sample1)
print("Second Sample:", sample2)
print("Are both samples equal?", np.array_equal(sample1, sample2))
```

3: Random sampling from an array

```python
# Select random elements from an array

students = np.array([
    "Aman", "Priya", "Rahul", "Neha", "Riya",
    "Karan", "Sneha", "Rohit", "Anjali", "Vikas"
])

# Select 4 students without replacement
random_students = np.random.choice(
    students, size=4, replace=False
)

print("All Students:", students)
print("Randomly Selected Students:", random_students)
```

4: Generate a simulated dataset

```python
# Generate random student marks between 40 and 100

np.random.seed(10)

marks = np.random.randint(40, 101, size=15)

print("Simulated Student Marks:", marks)
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

5: Random sampling and statistics

```python
# Select a random sample of 5 student marks

random_sample = np.random.choice(marks, size=5, replace=False)

print("Original Marks:", marks)
print("Random Sample:", random_sample)
print("Sample Mean:", np.mean(random_sample))
print("Sample Median:", np.median(random_sample))
print("Sample Standard Deviation:", np.std(random_sample))
```
