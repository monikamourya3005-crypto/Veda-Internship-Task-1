1: Import libraries and create dataset

```python

# Day 20: Data Types and Type Conversion
# Objective: Inspect and convert DataFrame columns to appropriate data types.

import pandas as pd

# Create a sample customer dataset with mixed data types.
df = pd.DataFrame({
    "CustomerID": ["101", "102", "103", "104", "105", "106"],
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Aman", "Neha"],
    "Age": ["25", "30", "22", "28", "24", "27"],
    "PurchaseAmount": ["2500.50", "4000.00", "1500.75",
                       "3200.25", "1800.00", "2750.50"],
    "IsMember": ["True", "False", "True", "False", "True", "False"],
    "JoinDate": ["2025-01-15", "2025-02-20", "2025-03-10",
                 "2025-04-05", "2025-05-12", "2025-06-18"]
})

print("Original Dataset:")
display(df)
```

2: Inspect data types

```python

# dtypes shows the data type of each column.
print("Data Types of Each Column:")
print(df.dtypes)

# info() provides column names, non-null counts, and data types.
print("\nDataset Information:")
df.info()

# Check the number of rows and columns.
print("\nDataset Shape:", df.shape)
```

3: Convert columns to numeric types

```python

# astype() converts a column to a specified data type.

# Convert CustomerID and Age from strings to integers.
df["CustomerID"] = df["CustomerID"].astype(int)
df["Age"] = df["Age"].astype(int)

# Convert PurchaseAmount from strings to floating-point numbers.
df["PurchaseAmount"] = pd.to_numeric(df["PurchaseAmount"])

print("After Numeric Conversion:")
display(df)

print("\nUpdated Data Types:")
print(df.dtypes)
```

4: Convert strings to booleans and dates

```python

# Convert text values into actual Boolean values.
# Explicit mapping avoids treating non-empty strings like "False" as True.
df["IsMember"] = df["IsMember"].map({
    "True": True,
    "False": False
}).astype(bool)

# Convert JoinDate from string to datetime.
df["JoinDate"] = pd.to_datetime(df["JoinDate"], errors="coerce")

print("After Boolean and Date Conversion:")
display(df)

print("\nUpdated Data Types:")
print(df.dtypes)
```

5: Handle invalid values safely

```python

# to_numeric(errors="coerce") converts invalid numeric values to NaN.
sample_values = pd.Series(["100", "250.5", "invalid", "400"])

converted_values = pd.to_numeric(sample_values, errors="coerce")

print("Original Values:")
display(sample_values)

print("Converted Numeric Values:")
display(converted_values)

# Check for missing or invalid values.
print("Missing or Invalid Values:")
print(converted_values.isna().sum())
```

6: Final cleaned dataset and validation

```python

# Create a cleaned copy with appropriate data types.
clean_df = df.copy()

print("Final Cleaned Dataset:")
display(clean_df)

print("Final Data Types:")
print(clean_df.dtypes)

# Verify the expected data types.
print("\nValidation Results:")
print("CustomerID is integer:", pd.api.types.is_integer_dtype(clean_df["CustomerID"]))
print("Age is integer:", pd.api.types.is_integer_dtype(clean_df["Age"]))
print("PurchaseAmount is numeric:", pd.api.types.is_numeric_dtype(clean_df["PurchaseAmount"]))
print("IsMember is Boolean:", pd.api.types.is_bool_dtype(clean_df["IsMember"]))
print("JoinDate is datetime:", pd.api.types.is_datetime64_any_dtype(clean_df["JoinDate"]))
```
