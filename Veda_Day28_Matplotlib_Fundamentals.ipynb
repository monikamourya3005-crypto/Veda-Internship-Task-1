1: Import Libraries

```python
# Day 28: Matplotlib Fundamentals
# Import libraries required for data visualization.

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

2: Create Dataset

```python
# Create a sample dataset for visualization.
# The dataset represents monthly sales and profit.

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [12000, 15000, 18000, 14000, 22000, 25000],
    "Profit": [2000, 3000, 4000, 2500, 5000, 6000],
    "Customers": [100, 120, 150, 130, 180, 200]
}

df = pd.DataFrame(data)

print(df)
```

3: Line Chart

```python
# A line chart shows trends over time.
# Here, we visualize monthly sales.

plt.figure(figsize=(8, 5))

plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.grid(True)
plt.show()
```

4: Bar Chart

```python
# A bar chart compares values across categories.
# Here, we compare monthly sales.

plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.show()
```

5: Scatter Plot

```python
# A scatter plot helps identify relationships between variables.
# Here, we compare sales and profit.

plt.figure(figsize=(8, 5))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

6: Histogram

```python
# A histogram shows the distribution of numerical data.
# Here, we visualize the distribution of sales.

plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=5, edgecolor="black")

plt.title("Distribution of Sales")
plt.xlabel("Sales Amount")
plt.ylabel("Frequency")

plt.show()
```

7: Multiple Charts in One Figure

```python
# Multiple charts help compare different aspects of a dataset.

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")

plt.subplot(1, 2, 2)
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")

plt.tight_layout()
plt.show()
```

8: Customize a Chart

```python
# Customize charts using labels, markers, and grid lines.
# Proper labels make visualizations easier to understand.

plt.figure(figsize=(8, 5))

plt.plot(
    df["Month"],
    df["Customers"],
    marker="o",
    linestyle="--",
    label="Customers"
)

plt.title("Monthly Customer Growth")
plt.xlabel("Month")
plt.ylabel("Number of Customers")

plt.legend()
plt.grid(True)

plt.show()
```

9: Save Chart

```python
# Save a visualization as an image file.
# The chart can be used in reports or presentations.

plt.figure(figsize=(8, 5))

plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("monthly_sales.png", dpi=300, bbox_inches="tight")

plt.show()

print("Chart saved successfully!")
```
