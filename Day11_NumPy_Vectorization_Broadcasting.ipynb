2 — Import NumPy and create arrays

```python
import numpy as np

# Create NumPy arrays
marks = np.array([65, 72, 80, 88, 95])
bonus = np.array([5, 3, 4, 2, 1])

print("Original Marks:", marks)
print("Bonus Marks:", bonus)
```

3 — Vectorized arithmetic operations

```python
# Perform operations without explicit Python loops

print("Addition:", marks + bonus)
print("Subtraction:", marks - bonus)
print("Multiplication:", marks * 2)
print("Division:", marks / 2)
print("Square of Marks:", marks ** 2)
```

4 — Scalar broadcasting

```python
# Adding a scalar to every array element

extra_marks = marks + 5

print("Original Marks:", marks)
print("Marks after adding 5:", extra_marks)
```

5 — Broadcasting between compatible shapes

```python
# A 2D array with 3 students and 3 subjects
student_marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 95]
])

# Add subject-wise bonus marks using broadcasting
subject_bonus = np.array([2, 3, 5])

updated_marks = student_marks + subject_bonus

print("Original Marks:\n", student_marks)
print("Subject Bonus:", subject_bonus)
print("Updated Marks:\n", updated_marks)
```

6 — Broadcasting with a column array

```python
# Add a different bonus to each student
student_bonus = np.array([[5], [3], [2]])

updated_student_marks = student_marks + student_bonus

print("Student-wise Bonus:\n", student_bonus)
print("Updated Marks:\n", updated_student_marks)
```

7 — Three broadcasting examples

```python
# Example 1: Array + scalar
a = np.array([10, 20, 30])
print("Example 1:", a + 5)

# Example 2: (3, 1) + (1, 3)
b = np.array([[1], [2], [3]])
c = np.array([[10, 20, 30]])
print("\nExample 2:\n", b + c)

# Example 3: (2, 3) + (3,)
d = np.array([[1, 2, 3], [4, 5, 6]])
e = np.array([10, 20, 30])
print("\nExample 3:\n", d + e)
```

8 — Compare loop-based and vectorized calculations

```python
import time

numbers = np.arange(1, 1_000_001)

# Loop-based calculation
start = time.perf_counter()

loop_result = []
for number in numbers:
    loop_result.append(number * 2)

loop_time = time.perf_counter() - start

# Vectorized calculation
start = time.perf_counter()

vectorized_result = numbers * 2

vectorized_time = time.perf_counter() - start

print(f"Loop-based time: {loop_time:.6f} seconds")
print(f"Vectorized time: {vectorized_time:.6f} seconds")
print("Results match:", np.array_equal(loop_result, vectorized_result))
```

9 — Student dataset analysis

```python
students = np.array([
    [78, 85, 90],
    [65, 70, 72],
    [88, 92, 95],
    [55, 60, 68],
    [91, 89, 94]
])

print("Student Marks:\n", students)
print("Subject-wise Average:", np.mean(students, axis=0))
print("Student-wise Average:", np.mean(students, axis=1))
print("Highest Marks:", np.max(students))
print("Lowest Marks:", np.min(students))
```
