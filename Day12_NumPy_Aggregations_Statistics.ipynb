1: Import NumPy and create a dataset

```python
import numpy as np

# Student performance dataset
data = np.array([
    [85, 90, 78],
    [92, 88, 95],
    [76, 80, 72],
    [89, 94, 86],
    [70, 75, 68],
    [95, 91, 98],
    [82, 85, 80],
    [88, 87, 90]
])

subjects = ["Python", "Mathematics", "Statistics"]

print("Student Performance Dataset:")
print(data)
print("\nDataset Shape:", data.shape)
print("Number of Students:", data.shape[0])
print("Number of Subjects:", data.shape[1])
```

2: Aggregation functions

```python
print("Total Marks:", np.sum(data))
print("Mean Marks:", np.mean(data))
print("Median Marks:", np.median(data))
print("Minimum Marks:", np.min(data))
print("Maximum Marks:", np.max(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
```

3: Row-wise and column-wise analysis

```python
# axis=0: Analyze each subject across all students
print("Subject-wise Mean:", np.mean(data, axis=0))
print("Subject-wise Minimum:", np.min(data, axis=0))
print("Subject-wise Maximum:", np.max(data, axis=0))

# axis=1: Analyze each student's marks
print("\nStudent-wise Mean:", np.mean(data, axis=1))
print("Student-wise Total:", np.sum(data, axis=1))
```

4: Display results in a readable format

```python
print("SUBJECT-WISE STATISTICAL ANALYSIS")
print("-" * 40)

for i, subject in enumerate(subjects):
    print(f"\n{subject}")
    print(f"Mean: {np.mean(data[:, i]):.2f}")
    print(f"Median: {np.median(data[:, i]):.2f}")
    print(f"Minimum: {np.min(data[:, i])}")
    print(f"Maximum: {np.max(data[:, i])}")
    print(f"Standard Deviation: {np.std(data[:, i]):.2f}")
    print(f"Variance: {np.var(data[:, i]):.2f}")
```

5: Interpretation of results

```python
subject_means = np.mean(data, axis=0)
best_subject = subjects[np.argmax(subject_means)]
lowest_subject = subjects[np.argmin(subject_means)]

print("RESULT INTERPRETATION")
print("-" * 40)
print(f"Overall Average Marks: {np.mean(data):.2f}")
print(f"Highest Average Subject: {best_subject}")
print(f"Lowest Average Subject: {lowest_subject}")
print(f"Highest Individual Mark: {np.max(data)}")
print(f"Lowest Individual Mark: {np.min(data)}")

print("\nConclusion:")
print(
    f"Students performed best on average in {best_subject}, "
    f"while {lowest_subject} had the lowest average. "
    "NumPy aggregation functions make statistical analysis "
    "of numerical datasets simple and efficient."
)
```
