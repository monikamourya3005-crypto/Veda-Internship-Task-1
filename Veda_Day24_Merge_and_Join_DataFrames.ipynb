1: Introduction and Import Libraries

```python
# Day 24: Merge and Join DataFrames
# Learn how to combine DataFrames using Pandas merge().

import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
```

2: Create Sample DataFrames

```python
# Create customer and order datasets
customers = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 104, 105],
    'Name': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'City': ['Indore', 'Bhopal', 'Mumbai', 'Delhi', 'Pune']
})

orders = pd.DataFrame({
    'Customer_ID': [101, 102, 103, 106, 107],
    'Order_ID': [501, 502, 503, 504, 505],
    'Amount': [2500, 1800, 3200, 1500, 2100]
})

print("Customer Data:")
display(customers)

print("Order Data:")
display(orders)
```

3: Inner Join

```python
# Inner join returns only matching Customer_ID records.

inner_join = pd.merge(
    customers, orders,
    on='Customer_ID',
    how='inner'
)

display(inner_join)
```

4: Left Join

```python
# Left join keeps all records from the left DataFrame.

left_join = pd.merge(
    customers, orders,
    on='Customer_ID',
    how='left'
)

display(left_join)
```

5: Right Join

```python
# Right join keeps all records from the right DataFrame.

right_join = pd.merge(
    customers, orders,
    on='Customer_ID',
    how='right'
)

display(right_join)
```

6: Outer Join

```python
# Outer join keeps all records from both DataFrames.

outer_join = pd.merge(
    customers, orders,
    on='Customer_ID',
    how='outer'
)

display(outer_join)
```

7: Compare Join Results

```python
# Compare the number of rows returned by different joins.

print("Inner Join Rows:", len(inner_join))
print("Left Join Rows:", len(left_join))
print("Right Join Rows:", len(right_join))
print("Outer Join Rows:", len(outer_join))
```

8: Identify Unmatched Records

```python
# Find customers who do not have matching orders.

unmatched_customers = left_join[
    left_join['Order_ID'].isna()
]

print("Customers without orders:")
display(unmatched_customers)

# Find orders without matching customer information.

unmatched_orders = right_join[
    right_join['Name'].isna()
]

print("Orders without customer information:")
display(unmatched_orders)
```

9: Analyze Combined Data

```python
# Calculate total order amount for each city.

city_sales = inner_join.groupby('City')['Amount'].sum()

print("Total Sales by City:")
display(city_sales)

# Calculate average order amount.

print("Average Order Amount:", inner_join['Amount'].mean())
```
