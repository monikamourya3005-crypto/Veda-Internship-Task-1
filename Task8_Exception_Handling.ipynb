**# Task 8: Exception Handling for Data Programs**

This notebook demonstrates how Python exception handling makes data-processing programs more reliable by handling invalid inputs and common runtime errors.

## 1. Common Python Exceptions

Explore common errors such as ValueError, TypeError, ZeroDivisionError, and IndexError.

```python
# ValueError
try:
    number = int("hello")
except ValueError:
    print("ValueError: Invalid number format")

# TypeError
try:
    result = "10" + 5
except TypeError:
    print("TypeError: Incompatible data types")

# ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError:
    print("ZeroDivisionError: Cannot divide by zero")

# IndexError
try:
    values = [10, 20, 30]
    print(values[5])
except IndexError:
    print("IndexError: List index out of range")
```

## 2. Using try, except, else, and finally

Demonstrate how Python handles successful operations, exceptions, and cleanup code.

```python
try:
    number = int(input("Enter a number: "))
    result = 100 / number

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Number cannot be zero.")

else:
    print("Result:", result)

finally:
    print("Operation completed.")
```

## 3. Safe Data Conversion

Convert input values to numbers while safely handling invalid data.

```python
def safe_convert(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        print("Invalid value:", repr(value))
        return None

data = ["25", "42.5", "hello", None, "18"]

converted_data = [safe_convert(item) for item in data]

print("Converted Data:", converted_data)
```

## 4. Exception Handling in Student Data

Validate student marks and handle invalid values without stopping the entire program.

```python
student_marks = ["85", "92", "invalid", "76", "-5", "101"]

valid_marks = []

for value in student_marks:
    try:
        mark = float(value)

        if not 0 <= mark <= 100:
            raise ValueError("Marks must be between 0 and 100")

    except ValueError as error:
        print(f"Skipping {value!r}: {error}")

    else:
        valid_marks.append(mark)

print("Valid Marks:", valid_marks)

if valid_marks:
    print("Average Marks:", sum(valid_marks) / len(valid_marks))
```

## 5. Handling Missing Customer Data

Process customer records while safely handling missing or invalid age values.

```python
customers = [
    {"name": "Aman", "age": "22"},
    {"name": "Priya", "age": "invalid"},
    {"name": "Ravi", "age": None},
    {"name": "Neha", "age": "25"}
]

for customer in customers:
    try:
        age = int(customer["age"])
    except (ValueError, TypeError):
        print(f"Skipping invalid age for {customer['name']}")
    else:
        print(f"{customer['name']} is {age} years old.")
```

## 6. Reusable Data-Processing Function

Create a reusable function that safely converts values and calculates their average.

```python
def calculate_average(values):
    valid_values = []

    for value in values:
        try:
            number = float(value)
        except (ValueError, TypeError):
            print("Skipping invalid value:", repr(value))
        else:
            valid_values.append(number)

    if not valid_values:
        return None

    return sum(valid_values) / len(valid_values)

data = ["10", "20", "invalid", "30", None]

average = calculate_average(data)
print("Average:", average)
```
