**# Task 3: Loops for Dataset Processing**

This notebook demonstrates how Python loops process collections of data and calculate summary statistics.

**## 1. Using a For Loop**

A for loop iterates over a sequence and processes each value.

```python
marks = [78, 85, 92, 67, 88]

for mark in marks:
    print("Student Marks:", mark)
```

**## 2. Using a While Loop**

A while loop repeats a block of code while a condition remains true.

```python
count = 1

while count <= 5:
    print("Iteration:", count)
    count += 1
```

**## 3. Processing a List of Numerical Values**

This program uses a loop to process numerical values and identify those greater than or equal to 50.

```python
values = [25, 60, 45, 80, 90, 35]

for value in values:
    if value >= 50:
        print(value, "is greater than or equal to 50")
```

**## 4. Calculating Summary Statistics**

This program uses loops and accumulators to calculate the total, average, minimum, and maximum values.

```python
sales = [1200, 1500, 1800, 1000, 2000]

total = 0
minimum = sales[0]
maximum = sales[0]

for sale in sales:
    total += sale

    if sale < minimum:
        minimum = sale

    if sale > maximum:
        maximum = sale

average = total / len(sales)

print("Sales:", sales)
print("Total:", total)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

**## 5. Iterating Through a Dictionary**

This example uses a loop to access keys and values in a dictionary.

```python
student_marks = {
    "Aman": 85,
    "Priya": 92,
    "Ravi": 76
}

for name, mark in student_marks.items():
    print(name, ":", mark)
```
