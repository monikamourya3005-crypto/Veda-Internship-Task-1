# Day 6 Task: List and Dictionary Comprehensions

```python
# Example 1: Basic List Comprehension - Squares of numbers from 1 to 10
squares = [x**2 for x in range(1, 11)]
print("Squares:", squares)
```

```python
# Example 2: Filtering with Condition - Even numbers from 1 to 20
even_numbers = [x for x in range(1, 21) if x % 2 == 0]
print("Even Numbers:", even_numbers)
```

```python
# Example 3: Transformation with String Manipulation - Convert names to uppercase
names = ["monika", "rahul", "priya", "amit"]
upper_names = [name.upper() for name in names]
print("Uppercase Names:", upper_names)
```

```python
# Example 4: Filtering and Transformation Combined - Square of even numbers only
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print("Even Squares:", even_squares)
```

```python
# Example 5: Student Marks Dataset - Conditional status assignment (Pass/Fail)
student_marks = [35, 78, 90, 25, 60, 45]
status = ["Pass" if mark >= 40 else "Fail" for mark in student_marks]
print("Student Status:", status)
```

```python
# Dictionary Example 1: Basic Dictionary Comprehension - Number and its cube
cube_dict = {x: x**3 for x in range(1, 6)}
print("Cube Dictionary:", cube_dict)
```

```python
# Dictionary Example 2: Filtering Existing Dictionary - Filter products with price > 500
product_prices = {"Laptop": 45000, "Mouse": 450, "Keyboard": 800, "Pen": 20}
expensive_products = {prod: price for prod, price in product_prices.items() if price > 500}
print("Expensive Products:", expensive_products)
```

```python
# Dictionary Example 3: Transformation - Converting student names and marks into grade dictionary
students = ["Monika", "Rahul", "Priya"]
marks = [85, 38, 92]
student_grades = {student: ("Pass" if m >= 40 else "Fail") for student, m in zip(students, marks)}
print("Student Grades:", student_grades)
```
