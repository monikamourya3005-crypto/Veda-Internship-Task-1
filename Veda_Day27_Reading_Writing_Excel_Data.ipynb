1: Import Libraries

```python
# Day 27: Reading and Writing Excel Data
# Learn how to read, process, and export Excel files.

import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

2: Create Sample Excel Dataset

```python
# Create a sample employee dataset and save it as an Excel file.

data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Amit', 'Priya', 'Rahul', 'Neha', 'Karan'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'HR'],
    'Salary': [45000, 35000, 55000, 40000, 30000],
    'Experience': [2, 1, 4, 3, 1]
}

df = pd.DataFrame(data)

df.to_excel('employees.xlsx', index=False)

print("Excel file created successfully!")
display(df)
```

3: Read Excel File

```python
# Read the Excel file using Pandas read_excel().

df = pd.read_excel('employees.xlsx')

print("Excel Data:")
display(df)
```

4: Inspect Excel Data

```python
# Explore the structure and information of the dataset.

print("First 5 rows:")
display(df.head())

print("Dataset Shape:", df.shape)

print("Column Names:")
print(df.columns)

print("Data Types:")
print(df.dtypes)

print("Dataset Information:")
df.info()
```

5: Check Missing Values

```python
# Identify missing values in the Excel dataset.

print("Missing Values:")
print(df.isnull().sum())

# Fill missing values if required.
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())

print("Missing values handled successfully!")
```

6: Filter Excel Data

```python
# Select employees working in the IT department.

it_employees = df[df['Department'] == 'IT']

print("IT Department Employees:")
display(it_employees)

# Select employees earning more than 40000.

high_salary = df[df['Salary'] > 40000]

print("Employees with Salary above 40000:")
display(high_salary)
```

7: Transform Data

```python
# Create a new column showing annual salary.

df['Annual_Salary'] = df['Salary'] * 12

# Create an experience category.

df['Experience_Level'] = np.where(
    df['Experience'] >= 3,
    'Experienced',
    'Fresher'
)

print("Transformed Dataset:")
display(df)
```

8: Sort Excel Data

```python
# Sort employees by salary in descending order.

sorted_df = df.sort_values(
    by='Salary',
    ascending=False
)

print("Employees Sorted by Salary:")
display(sorted_df)
```

9: Export Processed Data to Excel

```python
# Export the cleaned and transformed dataset.

df.to_excel('processed_employees.xlsx', index=False)

print("Processed Excel file saved successfully!")
```

10: Verify Exported File

```python
# Read the exported Excel file to verify the results.

final_df = pd.read_excel('processed_employees.xlsx')

print("Final Processed Excel Dataset:")
display(final_df)

print("Excel data processing completed successfully!")
```
