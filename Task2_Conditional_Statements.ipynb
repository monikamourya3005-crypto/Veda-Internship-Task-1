**# Task 2: Conditional Statements for Data Decisions**

This notebook demonstrates Python conditional statements through practical examples.

**## 1. Basic Conditional Examples**

This section covers 10 examples using if, elif, and else statements for everyday decisions.

```python
# Task 2: Conditional Statements for Data Decisions

# Example 1: Check age
age = 20
if age >= 18:
    print("Adult")
else:
    print("Minor")

# Example 2: Pass or Fail
marks = 75
if marks >= 40:
    print("Pass")
else:
    print("Fail")

# Example 3: Positive or Negative
number = -5
if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")

# Example 4: Even or Odd
number = 12
if number % 2 == 0:
    print("Even")
else:
    print("Odd")

# Example 5: Temperature
temperature = 35
if temperature > 30:
    print("Hot")
else:
    print("Normal")

# Example 6: Salary category
salary = 25000
if salary >= 50000:
    print("High Salary")
elif salary >= 25000:
    print("Medium Salary")
else:
    print("Low Salary")

# Example 7: Compare two numbers
a, b = 15, 10
if a > b:
    print("A is greater")
else:
    print("B is greater or equal")

# Example 8: Check score
score = 85
if score >= 90:
    print("Excellent")
elif score >= 75:
    print("Good")
elif score >= 40:
    print("Average")
else:
    print("Needs Improvement")

# Example 9: Check discount eligibility
purchase = 1200
if purchase >= 1000:
    print("Discount applicable")
else:
    print("No discount")

# Example 10: Check username
username = "Monika"
if username == "Monika":
    print("Welcome, Monika!")
else:
    print("Unknown user")
```

**## 2. Student Performance Classification**

This program classifies student performance based on marks using conditional statements.

```python
marks = 82

if marks >= 75:
    category = "High"
elif marks >= 50:
    category = "Medium"
elif marks >= 35:
    category = "Low"
else:
    category = "Fail"

print("Marks:", marks)
print("Performance Category:", category)
```

**## 3. Multiple Conditions Using AND and OR**

This section demonstrates how to combine conditions using logical operators to make decisions.

```python
age = 22
has_id = True

if age >= 18 and has_id:
    print("Eligible")
else:
    print("Not eligible")

attendance = 65
medical_leave = True

if attendance >= 75 or medical_leave:
    print("Condition satisfied")
else:
    print("Condition not satisfied")
```
