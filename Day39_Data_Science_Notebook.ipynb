2 – Import Libraries (Code cell)

```python
# Import required libraries for data analysis and visualization

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

3 – Load Dataset (Code cell)

```python
# Load the Titanic dataset

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

4 – Dataset Information (Code cell)

```python
# Understand the dataset structure

print("Dataset Information:")
df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Shape:")
print(df.shape)

print("\nFirst 5 Rows:")
display(df.head())
```

5 – Statistical Summary (Code cell)

```python
# Generate statistical summary of numerical columns

print("Statistical Summary:")
display(df.describe())

print("\nCategorical Summary:")
display(df.describe(include="object"))
```

6 – Check Missing Values (Code cell)

```python
# Identify missing values in each column

print("Missing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print((df.isnull().sum() / len(df)) * 100)
```

7 – Data Cleaning (Code cell)

```python
# Remove duplicate records

df = df.drop_duplicates()

# Fill missing Age values using median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values using mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Drop Cabin because it contains many missing values
df = df.drop(columns=["Cabin"])

print("Data cleaning completed!")

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

8 – Check Data Types (Code cell)

```python
# Check data types and convert where necessary

print("Data Types:")
print(df.dtypes)

df["Survived"] = df["Survived"].astype(int)

print("\nUpdated Data Types:")
print(df.dtypes)
```

9 – Exploratory Data Analysis (Code cell)

```python
# Explore important information from the dataset

print("Total Passengers:", len(df))

print("\nSurvival Count:")
print(df["Survived"].value_counts())

print("\nGender Distribution:")
print(df["Sex"].value_counts())

print("\nPassenger Class Distribution:")
print(df["Pclass"].value_counts())

print("\nAverage Age:", round(df["Age"].mean(), 2))

print("\nAverage Fare:", round(df["Fare"].mean(), 2))
```

10 – Survival Rate

```python
# Calculate overall survival rate

survival_rate = df["Survived"].mean() * 100

print("Overall Survival Rate:", round(survival_rate, 2), "%")

print("\nSurvival Distribution:")
display(df["Survived"].value_counts().rename({
    0: "Not Survived",
    1: "Survived"
}))
```

11 – Survival by Gender

```python
# Compare survival rates between male and female passengers

gender_survival = df.groupby("Sex")["Survived"].mean() * 100

print("Survival Rate by Gender:")
display(gender_survival)

gender_survival.plot(
    kind="bar",
    figsize=(7, 5),
    title="Survival Rate by Gender",
    xlabel="Gender",
    ylabel="Survival Rate (%)"
)

plt.xticks(rotation=0)
plt.show()
```

12 – Survival by Passenger Class

```python
# Analyze survival rates across passenger classes

class_survival = df.groupby("Pclass")["Survived"].mean() * 100

print("Survival Rate by Passenger Class:")
display(class_survival)

class_survival.plot(
    kind="bar",
    figsize=(7, 5),
    title="Survival Rate by Passenger Class",
    xlabel="Passenger Class",
    ylabel="Survival Rate (%)"
)

plt.xticks(rotation=0)
plt.show()
```

13 – Age Distribution

```python
# Visualize the age distribution of passengers

plt.figure(figsize=(8, 5))

sns.histplot(df["Age"], bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

14 – Survival Count Visualization

```python
# Visualize survived and non-survived passengers

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Survived")

plt.title("Passenger Survival Count")
plt.xlabel("Survival (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

15 – Correlation Heatmap

```python
# Analyze relationships between numerical variables

numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

16 – Key Findings and Conclusions

```python
# Generate conclusions from the analysis

print("KEY FINDINGS AND CONCLUSIONS")

print("\n1. Overall Survival Rate:",
      round(df["Survived"].mean() * 100, 2), "%")

print("\n2. Survival Rate by Gender:")
print((df.groupby("Sex")["Survived"].mean() * 100).round(2))

print("\n3. Survival Rate by Passenger Class:")
print((df.groupby("Pclass")["Survived"].mean() * 100).round(2))

print("\n4. Average Passenger Age:",
      round(df["Age"].mean(), 2))

print("\n5. Average Ticket Fare:",
      round(df["Fare"].mean(), 2))

print("\nConclusion:")
print("The analysis helps us understand passenger survival patterns.")
print("Gender, passenger class, and age are useful factors to explore")
print("when studying survival outcomes in the Titanic dataset.")
```

17 – Final Dataset Check

```python
# Display the cleaned dataset and save it

display(df.head(10))

print("Final Dataset Shape:", df.shape)
print("Remaining Missing Values:", df.isnull().sum().sum())

df.to_csv("Cleaned_Titanic_Dataset.csv", index=False)

print("\nCleaned dataset saved successfully!")
```
