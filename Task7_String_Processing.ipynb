**# Task 7: String Processing for Data Cleaning**

This notebook demonstrates Python string operations for cleaning and standardizing text data, including names, email addresses, and categories.

**## 1. Basic String Methods**

Practice trimming spaces, changing letter case, replacing characters, and splitting text.

```python
text = "  monika mourya  "

print("Original:", repr(text))
print("Trimmed:", text.strip())
print("Lowercase:", text.lower())
print("Uppercase:", text.upper())
print("Title Case:", text.strip().title())
print("Replace:", text.replace("monika", "Monika"))
print("Split:", text.strip().split())
```

**## 2. Cleaning Names**

Remove unwanted spaces and standardize names using title case.

```python
names = ["  monika mourya ", "RAHUL SHARMA  ", " priya  patel "]

clean_names = [name.strip().title() for name in names]

print("Original Names:", names)
print("Clean Names:", clean_names)
```

**## 3. Cleaning Email Addresses**

Remove extra spaces and convert email addresses to lowercase for consistent formatting.

```python
emails = [
    " Monika.Mourya@Gmail.com ",
    "RAHUL@EXAMPLE.COM",
    " priya@example.com "
]

clean_emails = [email.strip().lower() for email in emails]

print("Clean Emails:")
for email in clean_emails:
    print(email)
```

**## 4. Standardizing Categories**

Use string methods and replacement rules to standardize inconsistent category values.

```python
categories = [" data science ", "DATA SCIENCE", "Data-Science", " python ", "PYTHON"]

clean_categories = []

for category in categories:
    category = category.strip().lower().replace("-", " ")
    category = " ".join(category.split())
    clean_categories.append(category.title())

print("Original Categories:", categories)
print("Clean Categories:", clean_categories)
```

**## 5. Detecting Unwanted Spaces**

Identify strings containing leading or trailing spaces and extra spaces between words.

```python
values = ["Monika", " Rahul ", "Priya  Patel", "  Aman  Sharma  "]

for value in values:
    cleaned = " ".join(value.split())
    if value != cleaned:
        print(repr(value), "->", repr(cleaned))
```

**## 6. Reusable Text-Cleaning Function**

Create a function that removes extra spaces, standardizes letter case, and handles empty values.

```python
def clean_text(value):
    if not isinstance(value, str):
        return value
    return " ".join(value.split()).title()

sample_data = ["  monika mourya ", "RAHUL   SHARMA", "", " priya patel "]

for item in sample_data:
    print(repr(item), "->", repr(clean_text(item)))
```

**## 7. Customer Dataset Cleaning**

Apply string cleaning to customer names, emails, and categories to make the data more consistent.

```python
customers = [
    {"name": " monika mourya ", "email": " MONIKA@GMAIL.COM ", "category": " data science "},
    {"name": "RAHUL SHARMA ", "email": " rahul@example.com", "category": "PYTHON"},
    {"name": " priya  patel", "email": "PRIYA@EXAMPLE.COM ", "category": "Data-Science"}
]

for customer in customers:
    customer["name"] = clean_text(customer["name"])
    customer["email"] = customer["email"].strip().lower()
    customer["category"] = " ".join(
        customer["category"].strip().lower().replace("-", " ").split()
    ).title()

print("Cleaned Customer Dataset:")
for customer in customers:
    print(customer)
```
