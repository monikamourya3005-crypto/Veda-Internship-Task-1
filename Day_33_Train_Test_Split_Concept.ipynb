```python
# Day 33: Train-Test Split Concept
# Objective:
# Understand how to divide a dataset into training and testing sets.
# Training data is used to train the model.
# Testing data is used to evaluate model performance on unseen data.

print("Day 33 - Train-Test Split Concept")
print("Learning data splitting using Scikit-learn")
```

2: Import Libraries

```python
# Import required libraries

import pandas as pd
import numpy as np

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

print("Libraries imported successfully!")
```

3: Load Dataset

```python
# Load the Iris dataset

iris = load_iris()

X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target, name="Target")

print("Dataset loaded successfully!")
print("\nFeatures:")
display(X.head())

print("\nTarget values:")
print(y.head())
```

4: Dataset Information

```python
# Check dataset dimensions and target distribution

print("Dataset Shape:", X.shape)
print("\nTarget Distribution:")
print(y.value_counts())

print("\nFeature Information:")
print(X.info())
```

5: 80:20 Train-Test Split

```python
# Split dataset into 80% training and 20% testing data

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("80:20 Train-Test Split")
print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

70:30 Train-Test Split

```python
# Split dataset into 70% training and 30% testing data

X_train_70, X_test_30, y_train_70, y_test_30 = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("70:30 Train-Test Split")
print("Training Data:", X_train_70.shape)
print("Testing Data:", X_test_30.shape)
```

7: Compare Different Train-Test Ratios

```python
# Compare 80:20 and 70:30 splitting ratios

ratios = [0.20, 0.25, 0.30]

for ratio in ratios:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=ratio,
        random_state=42,
        stratify=y
    )

    print(f"\nTest Size: {int(ratio * 100)}%")
    print("Training Samples:", len(X_train))
    print("Testing Samples:", len(X_test))
```

8: Understand Random State

```python
# random_state ensures reproducibility
# Using the same random_state produces the same split.

X_train_1, X_test_1, y_train_1, y_test_1 = train_test_split(
    X, y, test_size=0.20, random_state=42
)

X_train_2, X_test_2, y_train_2, y_test_2 = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Are training datasets identical?")
print(X_train_1.equals(X_train_2))

print("\nAre testing datasets identical?")
print(X_test_1.equals(X_test_2))
```

9: Check Class Distribution

```python
# Stratified splitting maintains class proportions
# in both training and testing datasets.

print("Original Class Distribution:")
print(y.value_counts(normalize=True))

print("\nTraining Class Distribution:")
print(y_train_1.value_counts(normalize=True))

print("\nTesting Class Distribution:")
print(y_test_1.value_counts(normalize=True))
```
