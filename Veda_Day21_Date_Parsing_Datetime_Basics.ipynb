1: Import Libraries and Create Dataset

```python

# Day 21: Date Parsing and Datetime Basics
# Import required libraries

import pandas as pd
import numpy as np

# Create a sample Superstore-style sales dataset

data = {
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Customer": ["Aman", "Priya", "Rahul", "Sneha", "Neha",
                 "Ravi", "Anjali", "Karan", "Pooja", "Arjun"],
    "Order_Date": ["2025-01-15", "2025-02-20", "2025-03-10",
                   "2025-04-05", "2025-05-12", "2025-06-18",
                   "2025-07-22", "2025-08-25", "invalid", "2025-10-10"],
    "Sales": [2500, 4000, 1500, 3200, 1800, 2750, 2100, 3500, 2800, 4200]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

2: Inspect Dataset and Date Column

```python

# Check dataset information and column data types

print("Dataset Information:")
df.info()

print("\nDate Column:")
display(df["Order_Date"])

print("\nOriginal Data Type:")
print(df["Order_Date"].dtype)
```

3: Convert String Dates into Datetime

```python

# Convert text dates into Pandas datetime format.
# Invalid date values will become NaT.

df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], errors="coerce"
)

print("After Date Conversion:")
display(df)

print("\nUpdated Data Type:")
print(df["Order_Date"].dtype)
```

4: Identify and Handle Invalid Dates

```python

# Identify invalid or missing date values using isna()

invalid_dates = df[df["Order_Date"].isna()]

print("Invalid Date Records:")
display(invalid_dates)

print("Total Invalid Dates:", df["Order_Date"].isna().sum())

# Remove rows having invalid dates

df = df.dropna(subset=["Order_Date"]).copy()

print("\nDataset After Removing Invalid Dates:")
display(df)
```

5: Extract Year, Month and Day

```python

# Use the .dt accessor to extract date components

df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
df["Day"] = df["Order_Date"].dt.day

print("Dataset with Year, Month and Day:")
display(df)
```

6: Extract Weekday Information

```python

# Extract weekday name and weekday number

df["Weekday"] = df["Order_Date"].dt.day_name()
df["Weekday_Number"] = df["Order_Date"].dt.dayofweek

print("Dataset with Weekday Information:")
display(df[["Order_ID", "Order_Date", "Weekday", "Weekday_Number"]])
```

7: Date Conversion Examples

```python

# Convert dates written in different formats

date_values = pd.Series([
    "15-01-2025",
    "20/02/2025",
    "2025-03-10",
    "invalid_date"
])

converted_dates = pd.to_datetime(
    date_values, format="mixed",
    dayfirst=True, errors="coerce"
)

print("Original Dates:")
display(date_values)

print("Converted Datetime Values:")
display(converted_dates)
```

8: Filter Records Based on Dates

```python

# Select orders placed in the first quarter

quarter_one = df[df["Order_Date"].dt.quarter == 1]

print("First Quarter Orders:")
display(quarter_one)

# Select orders placed after June 2025

second_half = df[df["Order_Date"] >= "2025-07-01"]

print("\nOrders From July 2025 Onwards:")
display(second_half)
```

l 9: Date-Based Analysis

```python

# Analyze sales based on month and weekday

print("Total Orders:", len(df))
print("Total Sales:", df["Sales"].sum())
print("Average Sales:", round(df["Sales"].mean(), 2))

print("\nMonthly Sales:")
display(df.groupby("Month")["Sales"].sum())

print("\nWeekday-wise Sales:")
display(df.groupby("Weekday")["Sales"].sum())

print("\nYear-wise Orders:")
display(df.groupby("Year")["Order_ID"].count())
```
