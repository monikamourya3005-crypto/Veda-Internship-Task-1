1: Import NumPy

```python
import numpy as np

print("Day 10: NumPy Indexing and Slicing")
print("NumPy Version:", np.__version__)
```

2: One-Dimensional Array Indexing

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Array:", arr)
print("First Element:", arr[0])
print("Third Element:", arr[2])
print("Last Element:", arr[-1])
print("Second Last Element:", arr[-2])
```

3: Positive and Negative Indexing

```python
arr = np.array([5, 10, 15, 20, 25, 30])

print("First Element:", arr[0])
print("Fourth Element:", arr[3])
print("Element at Index -1:", arr[-1])
print("Element at Index -3:", arr[-3])
```

4: One-Dimensional Array Slicing

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])

print("Original Array:", arr)
print("First Three Elements:", arr[:3])
print("Elements from Index 2 to 5:", arr[2:6])
print("Elements from Index 3 Onwards:", arr[3:])
print("Every Second Element:", arr[::2])
print("Reversed Array:", arr[::-1])
```

5: Create a Two-Dimensional Array

```python
arr2 = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("2D Array:\n", arr2)
print("Element at Row 1, Column 2:", arr2[0, 1])
print("Element at Row 2, Column 3:", arr2[1, 2])
print("Last Element:", arr2[-1, -1])
```

6: Access Rows and Columns

```python
print("First Row:", arr2[0, :])
print("Second Row:", arr2[1, :])
print("First Column:", arr2[:, 0])
print("Second Column:", arr2[:, 1])
print("Last Column:", arr2[:, -1])
```

7: Two-Dimensional Slicing

```python
print("First Two Rows:\n", arr2[:2, :])
print("First Two Columns:\n", arr2[:, :2])
print("Top-Left 2x2 Subarray:\n", arr2[:2, :2])
print("Bottom-Right 2x2 Subarray:\n", arr2[1:, 1:])
```

8: Slicing with Steps

```python
arr = np.arange(1, 21)

print("Original Array:", arr)
print("Every Second Element:", arr[::2])
print("Every Third Element:", arr[::3])
print("Elements from Index 5 to 14:", arr[5:15])
print("Reverse Order:", arr[::-1])
```
