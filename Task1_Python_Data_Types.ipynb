**Task 1: Python for Data Science — Variables and Data Types**

**Objective**

To understand Python variables, built-in data types, type conversion, and basic arithmetic operations using real-world examples.


**1. Variables and Basic Data Types**

This section demonstrates integers, floating-point numbers, strings, and boolean values.

```python
# Variables and Data Types

name = "Monika"
age = 21
salary = 25000.50
is_student = True

print("Name:", name)
print("Age:", age)
print("Salary:", salary)
print("Is Student:", is_student)

print(type(name))
print(type(age))
print(type(salary))
print(type(is_student))
```

**2. Collection Data Types**

This section explains lists, tuples, sets, and dictionaries.

```python
#Collection data types
# List
marks = [80, 85, 90, 95]

# Tuple
coordinates = (10, 20)

# Set (stores unique values)
numbers = {1, 2, 2, 3, 4}

# Dictionary
student = {
    "name": "Monika",
    "age": 21,
    "course": "B.Tech AIML"
}

print("Marks:", marks)
print("Coordinates:", coordinates)
print("Unique numbers:", numbers)
print("Student details:", student)
```

**3. Type Conversion**

This section demonstrates converting strings into integer and float values.

```python
#Type conversion

age_text = "21"
salary_text = "25000.50"

age_number = int(age_text)
salary_number = float(salary_text)

print("Age:", age_number)
print("Salary:", salary_number)
print("Age data type:", type(age_number))
print("Salary data type:", type(salary_number))
```

**4. Basic Arithmetic Operations**

This section covers addition, subtraction, multiplication, division, and modulus.

```python
#Basic operations

a = 20
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Remainder:", a % b)
```

**5. Student Performance Analysis**

This section calculates total marks, average marks, and pass/fail status using Python

```python
#Real-world student performance example

student_marks = [78, 85, 92]

total = sum(student_marks)
average = total / len(student_marks)

print("Student Marks:", student_marks)
print("Total Marks:", total)
print("Average Marks:", average)

if average >= 40:
    print("Result: Pass")
else:
    print("Result: Fail")
```
