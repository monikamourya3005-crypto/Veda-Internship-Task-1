**# Task 5: Python Lists, Tuples, Sets and Dictionaries**

This notebook demonstrates Python collection data structures and their common operations through practical examples.

**## 1. Lists**

Lists are ordered and mutable collections that allow duplicate values. This example demonstrates adding, removing, searching, and sorting elements.

```python
# Creating a list
marks = [78, 85, 92, 67, 85]
print("Original List:", marks)

# Adding an element
marks.append(90)
print("After Append:", marks)

# Removing an element
marks.remove(67)
print("After Remove:", marks)

# Searching for an element
print("Is 85 present?", 85 in marks)

# Sorting the list
marks.sort()
print("Sorted List:", marks)
```

**## 2. Tuples**

Tuples are ordered and immutable collections. They are useful for storing values that should not be changed.

```python
student = ("Monika", 85, "AIML")

print("Student Tuple:", student)
print("Student Name:", student[0])
print("Marks:", student[1])
print("Number of Elements:", len(student))
print("Is 85 present?", 85 in student)
```

**## 3. Sets**

Sets store unique elements and support operations such as adding, removing, union, and intersection.

```python
set_a = {10, 20, 30, 40}
set_b = {30, 40, 50, 60}

set_a.add(50)
print("After Adding:", set_a)

set_a.discard(20)
print("After Removing:", set_a)

print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("Unique Values:", set(set_a))
```

**## 4. Dictionaries**

Dictionaries store data as key-value pairs. They allow values to be accessed, updated, added, and removed using keys.

```python
student = {
    "name": "Monika",
    "marks": 85,
    "department": "AIML"
}

print("Original Dictionary:", student)
print("Student Name:", student["name"])

# Updating a value
student["marks"] = 90

# Adding a new key-value pair
student["year"] = 3

# Removing a key-value pair
student.pop("department")

print("Updated Dictionary:", student)
```

**## 5. Comparison of Collection Types**

This section compares lists, tuples, sets, and dictionaries based on ordering, mutability, duplicates, and their common uses.

```python
comparison = {
    "List": ["Ordered", "Mutable", "Duplicates allowed"],
    "Tuple": ["Ordered", "Immutable", "Duplicates allowed"],
    "Set": ["Unordered", "Mutable", "Unique elements"],
    "Dictionary": ["Insertion-ordered", "Mutable", "Unique keys"]
}

for collection, properties in comparison.items():
    print(collection, ":", ", ".join(properties))
```

**## 6. Student Dataset Example**

This example combines Python collections to store student marks, identify unique scores, and access student records.

```python
students = [
    {"name": "Aman", "marks": 85},
    {"name": "Priya", "marks": 92},
    {"name": "Ravi", "marks": 78},
    {"name": "Neha", "marks": 85}
]

unique_marks = {student["marks"] for student in students}

print("Student Records:")
for student in students:
    print(student["name"], "-", student["marks"])

print("Unique Marks:", sorted(unique_marks))
print("Total Students:", len(students))
```
