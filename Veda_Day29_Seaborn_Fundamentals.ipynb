1: Import Libraries

```python
# Day 29: Seaborn Fundamentals
# Import libraries for statistical data visualization.

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

2: Create Dataset

```python
# Create a sample dataset for visualization.
# This dataset contains student marks and study hours.

data = {
    "Student": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Study_Hours": [2, 3, 4, 5, 6, 7, 8, 9],
    "Marks": [45, 50, 55, 65, 70, 75, 85, 90],
    "Department": ["AI", "CS", "AI", "CS", "AI", "CS", "AI", "CS"],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male"]
}

df = pd.DataFrame(data)

print(df)
```

3: Count Plot

```python
# Count plot displays the frequency of categorical values.
# Here, we count students in each department.

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Department")

plt.title("Student Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Students")

plt.show()
```

4: Box Plot

```python
# Box plot shows the distribution of numerical data.
# It helps identify median, spread, and potential outliers.

plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x="Department", y="Marks")

plt.title("Marks Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Marks")

plt.show()
```

5: Histogram

```python
# Histogram shows the distribution of numerical values.
# Here, we visualize the distribution of student marks.

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Marks", bins=5, kde=True)

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

6: Scatter Plot

```python
# Scatter plot shows the relationship between two numerical variables.
# Here, we explore study hours and marks.

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    hue="Department"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

7: Bar Plot

```python
# Bar plot compares numerical values across categories.
# Here, we compare average marks by department.

plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="Department", y="Marks")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")

plt.show()
```

8: Seaborn vs Matplotlib

```python
# Compare visualization using Seaborn and Matplotlib.
# Seaborn provides simple statistical plotting functions.

plt.figure(figsize=(12, 5))

# Matplotlib visualization
plt.subplot(1, 2, 1)
plt.scatter(df["Study_Hours"], df["Marks"])
plt.title("Matplotlib Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

# Seaborn visualization
plt.subplot(1, 2, 2)
sns.scatterplot(data=df, x="Study_Hours", y="Marks")
plt.title("Seaborn Scatter Plot")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.tight_layout()
plt.show()
```

9: Correlation Heatmap

```python
# Heatmap visualizes correlations between numerical variables.
# Correlation helps identify relationships between features.

plt.figure(figsize=(8, 5))

correlation = df[["Study_Hours", "Marks"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
