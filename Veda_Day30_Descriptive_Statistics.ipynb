1: Import Libraries

```python
# Import required libraries for statistical analysis

import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

2: Create Dataset

```python
# Create a sample dataset containing three numerical variables

data = {
    "Math_Score": [78, 85, 90, 65, 88, 92, 76, 81, 95, 70],
    "Science_Score": [82, 79, 91, 68, 85, 94, 73, 80, 89, 75],
    "Study_Hours": [2, 3, 4, 1, 3, 5, 2, 3, 5, 1]
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)
```

3: Basic Dataset Information

```python
# Display dataset information and numerical columns

print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nNumerical Columns:")
print(df.select_dtypes(include=np.number).columns.tolist())
```

4: Descriptive Statistics

```python
# Calculate descriptive statistics using describe()

print("Descriptive Statistics:")
display(df.describe())
```

5: Mean, Median and Mode

```python
# Calculate measures of central tendency

for column in df.columns:
    print(f"\n{column}")
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
```

6: Range, Variance and Standard Deviation

```python
# Calculate measures of data dispersion

for column in df.columns:
    data_range = df[column].max() - df[column].min()

    print(f"\n{column}")
    print("Range:", data_range)
    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())
```

7: Quartiles

```python
# Calculate Q1, Q2 and Q3 for each numerical column

for column in df.columns:
    print(f"\n{column}")
    print("Q1 (25%):", df[column].quantile(0.25))
    print("Q2 (50%):", df[column].quantile(0.50))
    print("Q3 (75%):", df[column].quantile(0.75))
```

8: Percentiles

```python
# Calculate selected percentiles

percentiles = [10, 25, 50, 75, 90]

for column in df.columns:
    print(f"\n{column} Percentiles:")

    for p in percentiles:
        value = np.percentile(df[column], p)
        print(f"{p}th Percentile: {value}")
```

9: Compare Mean and Median

```python
# Compare mean and median to understand data distribution

for column in df.columns:
    mean_value = df[column].mean()
    median_value = df[column].median()

    print(f"\n{column}")
    print("Mean:", mean_value)
    print("Median:", median_value)

    if mean_value > median_value:
        print("The data may be slightly right-skewed.")
    elif mean_value < median_value:
        print("The data may be slightly left-skewed.")
    else:
        print("Mean and median are equal.")
```

10: Final Statistical Summary

```python
# Display a final summary of statistical measures

summary = pd.DataFrame({
    "Mean": df.mean(),
    "Median": df.median(),
    "Minimum": df.min(),
    "Maximum": df.max(),
    "Range": df.max() - df.min(),
    "Variance": df.var(),
    "Standard Deviation": df.std()
})

print("Final Statistical Summary:")
display(summary)
```
