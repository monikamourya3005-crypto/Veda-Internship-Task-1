1: Import Libraries and Create Dataset

```python

# Day 23: Value Counts and Frequency Analysis
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create a sample customer dataset

data = {
    "Customer_ID": range(101, 121),
    "Gender": ["Male", "Female", "Female", "Male", "Male",
               "Female", "Male", "Female", "Male", "Female",
               "Male", "Female", "Female", "Male", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "City": ["Indore", "Bhopal", "Indore", "Khargone", "Bhopal",
             "Indore", "Delhi", "Bhopal", "Delhi", "Indore",
             "Khargone", "Delhi", "Bhopal", "Indore", "Delhi",
             "Khargone", "Indore", "Bhopal", "Delhi", "Indore"],

    "Category": ["Electronics", "Clothing", "Food", "Electronics",
                 "Food", "Clothing", "Electronics", "Food",
                 "Clothing", "Electronics", "Food", "Clothing",
                 "Electronics", "Food", "Clothing", "Electronics",
                 "Food", "Clothing", "Electronics", "Food"],

    "Payment_Method": ["UPI", "Cash", "Card", "UPI", "Cash",
                       "Card", "UPI", "Cash", "Card", "UPI",
                       "Cash", "Card", "UPI", "Cash", "Card",
                       "UPI", "Cash", "Card", "UPI", "Cash"],

    "Membership": ["Gold", "Silver", "Regular", "Gold", "Silver",
                   "Regular", "Gold", "Regular", "Silver", "Gold",
                   "Regular", "Silver", "Gold", "Regular", "Silver",
                   "Gold", "Regular", "Silver", "Gold", "Regular"]
}

df = pd.DataFrame(data)

print("Customer Dataset:")
display(df)
```

2: Dataset Information

```python

# Display basic information about the dataset

print("Dataset Shape:", df.shape)

print("\nColumn Information:")
df.info()

print("\nFirst Five Records:")
display(df.head())

# Identify categorical columns

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("\nCategorical Columns:")
print(categorical_columns)
```

3: Frequency Analysis of Gender

```python

# value_counts() counts occurrences of each category.

gender_counts = df["Gender"].value_counts()

print("Gender Frequency:")
display(gender_counts)

# Calculate percentage distribution

gender_percentage = df["Gender"].value_counts(normalize=True) * 100

print("\nGender Percentage Distribution:")
display(gender_percentage.round(2))
```

4: Frequency Analysis of City

```python

# Analyze customer distribution across different cities.

city_counts = df["City"].value_counts()

print("City Frequency:")
display(city_counts)

# Calculate city-wise proportions

city_percentage = df["City"].value_counts(normalize=True) * 100

print("\nCity Percentage Distribution:")
display(city_percentage.round(2))
```

5: Frequency Analysis of Product Category

```python

# Count the number of purchases in each product category.

category_counts = df["Category"].value_counts()

print("Product Category Frequency:")
display(category_counts)

# Calculate percentage distribution

category_percentage = (
    df["Category"].value_counts(normalize=True) * 100
)

print("\nProduct Category Percentage:")
display(category_percentage.round(2))
```

6: Frequency Analysis of Payment Method

```python

# Analyze customer payment preferences.

payment_counts = df["Payment_Method"].value_counts()

print("Payment Method Frequency:")
display(payment_counts)

# Calculate payment method proportions

payment_percentage = (
    df["Payment_Method"].value_counts(normalize=True) * 100
)

print("\nPayment Method Percentage:")
display(payment_percentage.round(2))
```

7: Frequency Analysis of Membership

```python

# Analyze different customer membership types.

membership_counts = df["Membership"].value_counts()

print("Membership Frequency:")
display(membership_counts)

# Calculate membership percentage

membership_percentage = (
    df["Membership"].value_counts(normalize=True) * 100
)

print("\nMembership Percentage:")
display(membership_percentage.round(2))
```

8: Combined Frequency Analysis

```python

# Perform frequency analysis on all five categorical columns.

for column in categorical_columns:

    print("\nFrequency Analysis of:", column)

    frequency = df[column].value_counts()
    percentage = df[column].value_counts(normalize=True) * 100

    result = pd.DataFrame({
        "Frequency": frequency,
        "Percentage": percentage.round(2)
    })

    display(result)
```

9: Visualize Categorical Distributions

```python

# Create bar charts for categorical columns.

for column in categorical_columns:

    plt.figure(figsize=(8, 4))

    sns.countplot(
        data=df,
        x=column,
        order=df[column].value_counts().index
    )

    plt.title(f"Frequency Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
```

10: Interpretation of Results

```python

# Identify the most frequent category in each column.

print("DATA FREQUENCY ANALYSIS - INTERPRETATION\n")

for column in categorical_columns:

    counts = df[column].value_counts()
    percentages = df[column].value_counts(normalize=True) * 100

    most_common = counts.idxmax()
    highest_count = counts.max()
    highest_percentage = percentages.max()

    print(f"Column: {column}")
    print(f"Most Frequent Category: {most_common}")
    print(f"Frequency: {highest_count}")
    print(f"Percentage: {highest_percentage:.2f}%")
    print("-" * 40)

# Final conclusion

print("\nConclusion:")
print("Value counts help identify the frequency of categorical values.")
print("Percentage distribution helps compare categories.")
print("These techniques are useful for understanding dataset patterns.")
```
