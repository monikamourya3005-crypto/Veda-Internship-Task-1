1: Import NumPy

```python
import numpy as np

print("NumPy Version:", np.__version__)
print("Day 9 Task: NumPy Array Creation and Data Types")
```

2: Create a 1D Array

```python
arr1 = np.array([10, 20, 30, 40, 50])

print("1D Array:", arr1)
print("Number of Dimensions:", arr1.ndim)
print("Shape:", arr1.shape)
print("Size:", arr1.size)
print("Data Type:", arr1.dtype)
```

3: Create a 2D Array

```python
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("2D Array:\n", arr2)
print("Number of Dimensions:", arr2.ndim)
print("Shape:", arr2.shape)
print("Size:", arr2.size)
print("Data Type:", arr2.dtype)
```

4: Explore Different Data Types

```python
int_arr = np.array([1, 2, 3], dtype=np.int32)
float_arr = np.array([1.5, 2.5, 3.5], dtype=np.float64)
bool_arr = np.array([True, False, True])

print("Integer Array:", int_arr, "| Type:", int_arr.dtype)
print("Float Array:", float_arr, "| Type:", float_arr.dtype)
print("Boolean Array:", bool_arr, "| Type:", bool_arr.dtype)
```

5: Create Arrays Using NumPy Functions

```python
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((2, 3))
range_arr = np.arange(1, 11)
line_arr = np.linspace(0, 10, 5)

print("Zeros Array:\n", zeros_arr)
print("\nOnes Array:\n", ones_arr)
print("\nRange Array:", range_arr)
print("\nEvenly Spaced Values:", line_arr)
```

6: Perform Numerical Operations

```python
a = np.array([10, 20, 30, 40])
b = np.array([1, 2, 3, 4])

print("Array A:", a)
print("Array B:", b)
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

7: Compare NumPy Arrays with Python Lists

```python
python_list = [1, 2, 3, 4]
numpy_array = np.array([1, 2, 3, 4])

print("Python List:", python_list)
print("NumPy Array:", numpy_array)

print("List Repetition:", python_list * 2)
print("Array Multiplication:", numpy_array * 2)
```

8: Student Marks Dataset

```python
marks = np.array([[85, 90, 78],
                  [70, 88, 92],
                  [95, 82, 89]])

print("Student Marks:\n", marks)
print("Shape:", marks.shape)
print("Total Students:", marks.shape[0])
print("Subjects per Student:", marks.shape[1])
print("Average Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

9: Iris-Style Numerical Dataset

```python
iris_sample = np.array([
    [5.1, 3.5, 1.4, 0.2],
    [4.9, 3.0, 1.4, 0.2],
    [6.3, 3.3, 6.0, 2.5]
])

print("Sample Dataset:\n", iris_sample)
print("Shape:", iris_sample.shape)
print("Dimensions:", iris_sample.ndim)
print("Total Elements:", iris_sample.size)
print("Column-wise Mean:", np.mean(iris_sample, axis=0))
```
